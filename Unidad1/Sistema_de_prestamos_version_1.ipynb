```python
# =============================================
# SISTEMA DE PRESTAMOS - VERSION1
# =============================================

# Lista de alumnos activos
alumnos = ["26040200", "26040201", "26040202", "26040203"]

# Inventario de computadoras y sus atributos
computadoras = [
    "PC-LC1-01", "PC-LC1-02", "PC-LC1-03",
    "PC-LC2-01", "PC-LC2-02", "PC-LC2-03",
    "PC-LC3-01", "PC-LC3-02", "PC-LC3-03", "PC-LC3-04"
]

aulas = [
    "LC1", "LC1", "LC1",
    "LC2", "LC2", "LC2",
    "LC3", "LC3", "LC3", "LC3"
]

software = [
    "Python, VS code", "Python, Java, VS code", "autoCAD, MATLAB",
    "Python, VS code", "Python, Java, VS code", "autoCAD, MATLAB",
    "Python, VS code", "Python, Java, VS code", "autoCAD, MATLAB", "Python, VS code"
]

estados = [
    "Disponible", "Prestada", "Reservada",
    "Disponible", "Prestada", "Reservada",
    "Disponible", "Prestada", "Reservada", "Mantenimiento"
]

opcion = 0

while opcion != 7:
    print("\n=====================================")
    print("SISTEMA DE PRESTAMOS")
    print("=====================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar prestamo")
    print("4. Devolver computadora")
    print("5. Crear reservacion")
    print("6. Agregar computadora")
    print("7. Salir")
    print("=====================================")

    opcion = int(input("Seleccione una opcion: "))

    # 1. Consultar computadoras
    if opcion == 1:
        print("\n--- COMPUTADORAS ---")
        for i in range(len(computadoras)):
            print(f"Aula: {aulas[i]},ID: ,{computadoras[i]},Software: ,{software[i]} ,Estado: ,{estados[i]}")

    # 2. Buscar por software
    elif opcion == 2:
        programa = input("\nIngrese el software que desea buscar: ")
        print("\n--- RESULTADOS ---")
        encontrado = False
        for i in range(len(computadoras)):
            if programa.lower() in software[i].lower():
                print(f"ID: {computadoras[i]} | Aula: {aulas[i]} | Software: {software[i]} | Estado: {estados[i]}")
                encontrado = True
        if not encontrado:
            print("No se encontraron computadoras con ese software.")

    # 3. Solicitar prestamo
    elif opcion == 3:
        print("\n--- SOLICITAR PRESTAMO ---")
        num_control = input("Ingrese el numero de control del alumno: ")

        if num_control in alumnos:
            id_pc = input("Ingrese el ID de la computadora: ")
            encontrado_pc = False

            for i in range(len(computadoras)):
                if computadoras[i] == id_pc:
                    encontrado_pc = True
                    if estados[i] == "Disponible":
                        estados[i] = "Prestada"
                        print(f"¡Prestamo registrado! Se asignó la computadora {computadoras[i]} del aula {aulas[i]}.")
                    else:
                        print(f"La computadora no esta disponible (Estado actual: {estados[i]}).")
                    break

            if not encontrado_pc:
                print("La computadora ingresada no existe.")
        else:
            print("El alumno no está registrado en el sistema.")

    # 4. Devolver computadora
    elif opcion == 4:
        print("\n--- DEVOLVER COMPUTADORA ---")
        id_pc = input("Ingrese el ID de la computadora a devolver: ")
        encontrado_pc = False

        for i in range(len(computadoras)):
            if computadoras[i] == id_pc:
                encontrado_pc = True
                if estados[i].lower() in ["prestada", "reservada"]:
                    estados[i] = "Disponible"
                    print(f"La computadora {id_pc} ha sido devuelta y ahora esta Disponible.")
                else:
                    print(f"La computadora {id_pc} no estaba prestada ni reservada (Estado: {estados[i]}).")
                break

        if not encontrado_pc:
            print("La computadora ingresada no existe.")

    # 6. Agregar computadora
    elif opcion == 6:
        print("\n--- AGREGAR COMPUTADORA ---")
        id_pc = input("ID de la nueva computadora: ")
        aula = input("Aula: ")
        programas = input("Software instalado: ")

        computadoras.append(id_pc)
        aulas.append(aula)
        software.append(programas)
        estados.append("Disponible")
        print("Computadora agregada correctamente.")

    elif opcion == 7:
        print("\nSaliendo del sistema...")
```
