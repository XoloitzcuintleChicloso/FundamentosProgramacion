```python
def tablas_de_multiplicar(x):
  print("TABLA DEL ", x)
  print("=============")
  for i in range(10):
    print(x, "por", i, "=", x*i)
  print()
```

```python
tablas_de_multiplicar(3)
tablas_de_multiplicar(5)
```

## *Regresar un resultado - La intrucción return*
- Si queremos que el resultado del proceso de la funcion se le asigne a una variable, debemos usar la palabra reservada return.


```python
# Definir una funcion que calcule el area de un triangulo
# la funcion recibe 2 parametros, la base y la altura

def calcular_area(base, altura):
  area = base * altura
  return area
```

```python
# Primer ejemplo de como se puede hacer:
resultado = calcular_area(26,14)
print("Area del triangulo", resultado)

# Segundo ejemplo:
print("Area del triangulo", calcular_area(10,5))
```

```python
# - TODAS las variables que declaramos dentro de una funcion
# existen solo dentro de la funcion, esto es una funcion local

# - Si queremos usarlas en el resto del codigo debemos declararlas como
# variables globales
```

```python
def miFuncion():
  variablex = 123  # Es una variable local
  print(variablex)
# La variable x de miFuncion existe unicamente dentro del
# codigo de la funcion

miFuncion()

# La siguiente variable NO es la misma que esta definida en miFuncion,
# por lo tanto, no existe la variabe y por eso marca error

print(variablex)
```

```python
def miFuncion():
  global variablex  # Este codigo vuelve variablex una variable global
  variablex = 123
  print(variablex)

miFuncion()
print(variablex)
```
